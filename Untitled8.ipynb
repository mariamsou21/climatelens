```python
%%writefile fetcher.py
import os, time, requests

BASE = "https://climate.weather.gc.ca/climate_data/bulk_data_e.html"

def download_year(station, year, cache_dir="cache", refresh=False, retries=3):
    os.makedirs(cache_dir, exist_ok=True)
    path = os.path.join(cache_dir, f"{station}_{year}.csv")
    if os.path.exists(path) and not refresh:
        return path
    params = {"format": "csv", "stationID": station, "Year": year,
              "Month": 1, "Day": 1, "timeframe": 2, "submit": "Download Data"}
    for attempt in range(1, retries + 1):
        try:
            r = requests.get(BASE, params=params, timeout=30)
            r.raise_for_status()
            if len(r.content) < 200:
                raise ValueError("empty file: no data for this station/year")
            with open(path, "wb") as f:
                f.write(r.content)
            return path
        except (requests.RequestException, ValueError):
            if attempt == retries:
                raise
            time.sleep(2 * attempt)

def download_years(station, start, end, **kw):
    return [download_year(station, y, **kw) for y in range(start, end + 1)]
```

```python
%%writefile cleaner.py
import pandas as pd

# Columns I keep from the ECCC file, with short names for my own code
COLUMNS = {
    "Date/Time": "date",
    "Max Temp (°C)": "tmax",
    "Min Temp (°C)": "tmin",
    "Mean Temp (°C)": "tmean",
    "Total Precip (mm)": "precip",}

# ECCC quality-flag columns (M = missing, E = estimated, A = accumulated, T = trace)
FLAGS = {
    "Max Temp Flag": "tmax_flag",
    "Min Temp Flag": "tmin_flag",
    "Mean Temp Flag": "tmean_flag",
    "Total Precip Flag": "precip_flag",}


def read_year(path):
    """Read one cached ECCC file (one station, one year) and return a clean table."""
    # utf-8-sig removes the invisible byte-order mark at the start of the file
    df = pd.read_csv(path, encoding="utf-8-sig")

    # Stop with a clear message if ECCC renamed or removed a column I need
    missing = [c for c in COLUMNS if c not in df.columns]
    if missing:
        raise ValueError(f"{path}: missing columns {missing}")

    # Keep only the columns I need and give them short names
    out = df[list(COLUMNS)].rename(columns=COLUMNS)

    # Copy each flag column as text; use empty text if the column is absent
    for old, new in FLAGS.items():
        empty = pd.Series(pd.NA, index=df.index, dtype="string")
        out[new] = df[old].astype("string") if old in df.columns else empty

    # Turn the date text into real dates
    out["date"] = pd.to_datetime(out["date"])

    # Turn measurements into numbers; anything unreadable becomes NaN (a gap)
    for c in ["tmax", "tmin", "tmean", "precip"]:
        out[c] = pd.to_numeric(out[c], errors="coerce")

    # Trace precipitation (flag T) has no number: count it as 0.0, keep the flag
    trace = (out["precip_flag"] == "T").fillna(False) & out["precip"].isna()
    out.loc[trace, "precip"] = 0.0
    return out


def load_station(paths):
    """Read several yearly files and join them into one table, sorted by date."""
    frames = [read_year(p) for p in paths]
    data = pd.concat(frames, ignore_index=True)
    return data.sort_values("date").set_index("date")


def quality_report(df, min_complete=90):
    """Per year: days, % of days with a value, flagged values, and pass/fail."""
    g = df.groupby(df.index.year)
    rep = pd.DataFrame({
        "days": g["tmean"].size(),
        "tmean_%": (g["tmean"].count() / g["tmean"].size() * 100).round(1),
        "precip_%": (g["precip"].count() / g["precip"].size() * 100).round(1),})
    # Count flagged values (non-empty flag cells) for each year
    flag_cols = [c for c in df.columns if c.endswith("_flag")]
    rep["flagged"] = df[flag_cols].notna().sum(axis=1).groupby(df.index.year).sum()
    # True if the year has enough data to be trusted
    rep["kept"] = rep["tmean_%"] >= min_complete
    return rep
```

```python
%%writefile test_pipeline.py
import cleaner
import fetcher


def make_csv(tmp_path, rows, name="2024.csv"):
    """Make a small fake ECCC file (with the invisible byte-order mark, like the real ones)."""
    header = "Date/Time,Max Temp (°C),Min Temp (°C),Mean Temp (°C),Total Precip (mm),Total Precip Flag\n"
    p = tmp_path / name
    p.write_text(header + "\n".join(rows) + "\n", encoding="utf-8-sig")
    return str(p)


def test_read_year_basic(tmp_path):
    # 3 days; day 2 has no mean temperature
    path = make_csv(tmp_path, [
        "2024-01-01,1.0,-3.0,-1.0,0.5,",
        "2024-01-02,2.0,-2.0,,,",
        "2024-01-03,3.0,-1.0,1.0,0.0,",
    ])
    df = cleaner.read_year(path)
    assert len(df) == 3                      # all rows kept
    assert df["tmean"].count() == 2          # one gap stays a gap (not filled in)
    assert str(df["precip_flag"].dtype) == "string"   # flags are text (issue #2)


def test_trace_precip_counts_as_zero(tmp_path):
    # no number but flag T (trace) -> should become 0.0, flag kept
    path = make_csv(tmp_path, ["2024-01-01,0.0,-4.0,-2.0,,T"])
    df = cleaner.read_year(path)
    assert df.loc[0, "precip"] == 0.0
    assert df.loc[0, "precip_flag"] == "T"


def test_missing_column_gives_clear_error(tmp_path):
    # a file without the "Max Temp" column must stop with a clear message
    p = tmp_path / "bad.csv"
    p.write_text("Date/Time,Mean Temp (°C)\n2024-01-01,1.0\n", encoding="utf-8-sig")
    try:
        cleaner.read_year(str(p))
        assert False, "should have raised ValueError"
    except ValueError as e:
        assert "missing columns" in str(e)


def test_quality_report_90_percent_rule(tmp_path):
    # 10 days, 1 missing -> 90 % -> kept
    rows = [f"2024-01-{d:02d},1.0,-3.0,-1.0,0.5," for d in range(1, 10)]
    rows.append("2024-01-10,1.0,-3.0,,0.5,")
    rep = cleaner.quality_report(cleaner.load_station([make_csv(tmp_path, rows)]))
    assert rep.loc[2024, "tmean_%"] == 90.0
    assert bool(rep.loc[2024, "kept"]) is True
    # 10 days, 2 missing -> 80 % -> not kept
    rows[8] = "2024-01-09,1.0,-3.0,,0.5,"
    rep = cleaner.quality_report(cleaner.load_station([make_csv(tmp_path, rows, "b.csv")]))
    assert bool(rep.loc[2024, "kept"]) is False


class FakeResponse:
    """Pretends to be the ECCC website's answer (no internet needed in tests)."""
    def __init__(self, size):
        self.content = b"x" * size
    def raise_for_status(self):
        pass


def test_fetcher_uses_cache(tmp_path, monkeypatch):
    calls = []
    monkeypatch.setattr(fetcher.requests, "get", lambda *a, **k: calls.append(1) or FakeResponse(300))
    cache = str(tmp_path / "cache")
    fetcher.download_year(4333, 2024, cache_dir=cache)   # first time: downloads
    fetcher.download_year(4333, 2024, cache_dir=cache)   # second time: uses the saved file
    assert len(calls) == 1
    fetcher.download_year(4333, 2024, cache_dir=cache, refresh=True)  # refresh: downloads again
    assert len(calls) == 2


def test_fetcher_empty_file_raises(tmp_path, monkeypatch):
    monkeypatch.setattr(fetcher.requests, "get", lambda *a, **k: FakeResponse(10))
    monkeypatch.setattr(fetcher.time, "sleep", lambda s: None)   # do not really wait
    try:
        fetcher.download_year(4333, 1800, cache_dir=str(tmp_path / "c"))
        assert False, "should have raised ValueError"
    except ValueError as e:
        assert "empty file" in str(e)
```
